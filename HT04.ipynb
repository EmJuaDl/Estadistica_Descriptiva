### BLOQUE 1

**Pregunta 1: ¿Cómo ha evolucionado el rendimiento promedio de combustible a lo largo de los años del modelo?**

Tipo de análisis: Tendencia temporal

Gráfico generado: 

```python
plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='model_year', y='mpg', marker='o', errorbar=('ci', 95))
plt.title('Evolución del Rendimiento Promedio de Combustible (1970 - 1982)')
plt.xlabel('Año del Modelo (1970-1982)')
plt.ylabel('Rendimiento (mpg)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

¿La tendencia es creciente, decreciente o estable?
-> La tendencia es creciente. A lo largo del tiempo, podemos ver que el rendimiento pasa de 18 mpg a cerca de 31 mpg.

¿Existe algún año donde se rompe el patrón?
-> Sí, entre 1971 y 1973 podemos observar un leve descenso, donde el rendimiento disminuye previo de haber tenido una subida. De igual manera en 1975  y en 1981, donde cae. 

¿Qué magnitud tiene el cambio entre el primer y último año (en mpg y en %)?
En mpg, pasa de 17.7 mpg en 1970 a 31.7 mpg en 1982. Un cambio de 14 mpg, lo cual es aproximadamente un 79.1%.

¿Qué factores ingenieriles podrían explicar esta tendencia?
La crisis del petróleo de la década de 1970, la cual impulsó regulaciones gubernamentales de eficiencia. Forzó el diseño de motores más pequeños, reducción del peso del vehículo, entre otras. 

**Interpretación técnica:** 
El gráfico nos muestra un patrón de agrupamiento donde los vehículos más lentos tienden a mostrar cifras superiores de mpg. Incorporar el mapa de color revela que la potencia actúa como determinante del sietema, ya que los motores con alta potencia logran aceleraciones más rápidas a costa de un mayor deterioro. Mientras los vehículos eficiente sacrifican aceleración por kilometraje.

¿Sería apropiado usar la media como medida representativa? Justifica.

La variable horsepower muestra ua asimetría positiva clara, donde el 50% central de la muestra se agrupa entre 75 y 125 hp, presenta 6 valores atípicos en el extremo superior que superan los 200 hp. **Pregunta 3: ¿Cómo se relacionan el peso del vehículo y el rendimiento de combustible?**

Tipo de análisis: Relación entre dos variables cuantitativas

Gráfico:

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="weight", y="mpg", hue="origin", alpha=0.7, s=60)
sns.regplot(data=df,x="weight",y="mpg",scatter=False,color="black",line_kws={"linestyle": "--", "linewidth": 2},)

plt.title("Relación entre Peso del Vehículo y Rendimiento de Combustible")
plt.xlabel("Peso del Vehículo (lb)")
plt.ylabel("Rendimiento de Combustible (mpg)")
plt.legend(title="Origen")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

¿La relación es positiva o negativa? ¿Qué implica físicamente?
-> Es negativa, ya que es inversa, entre mayor el peso del vehículo, menor el rendimiento.
    
¿La relación parece lineal? ¿Se observa algún patrón curvilíneo?
-> Sí, la relación parece lineal, no se observa algún patrón curvilíneo. 
    
¿Los vehículos de distintos orígenes se comportan igual?
-> No, los vehículos japoneses no superan las 3000 lb, al igual que los europeos que casi en su totalidad en la zona de bajo peso. Mientras que los vehículos estadounidenses se concentran en la zona de mayor peso.

Estima visualmente el coeficiente de correlación de Pearson (r).
-> Visualmente es de -0,9 el coeficiente de Pearson.
    
¿Qué implicaciones tiene esto para el diseño de motores más eficientes?
-> Implica que para incrementar el rendimiento se debe tomar en cuenta reducir el peso del vehículo.

**Interpretación técnica:**
El análisis de dispersión entre peso del vehículo y el rendimiento evidencia un acoplamiento físico lineal e inversamente proporcional respaldado por una marcada correlación negativa. La línea de tendencia captura la trayectoria promedio del consumo energético del vehículo respecto a su masa. Asimismo, la división según el origen muestra la marcada diferencia en cuanto a fabricación.

-> El grupo de 4 cilindros presenta mayor dispersión, podría deberse a que engloba una gran variedad de diseños, tecnologías de motor, pesos del vehículo y años de fabricación.

¿Hay solapamiento entre las cajas? ¿Qué implica estadísticamente?
-> Sí, la caja de 5 cilindros se solapa ampliamente con la de 4 cilindros, mientras que la de 3 cilindros se solapa casi por completo con la caja de 6 cilindros.

Que las cajas centrales de 4, 6 y 8 cilindros no se solapen significa que cada grupo tiene un rendimiento claramente distinto. Aunque los extremos se crucen en algunos casos, la diferencia general es tan grande que la prueba ANOVA confirmará que el número de cilindros sí influye de forma significativa en el consumo de combustible.

Formula una hipótesis que luego podrías contrastar con ANOVA.
-> Al menos uno de los grupos de cilindros tiene un rendimiento promedio de combustible diferente a los demás.

**Interpretación técnica:**
El diagrama de caja relfeja una relación inversamente proporcional entre el número de cilindros y el rendimiento de combustible. Las distribuciones centrales de las categorías mayoritarias están claramente estratificadas, demostrando que la mediana de rendimiento cae de forma escalonada. Además, la marcada separación entre bloques centrales confirma que el número de cilindros es un factor explicativo para la variabilidad en mpg.

**Pregunta 8: ¿Cómo ha cambiado la composición de cilindros y orígenes por año?**

Tipo de análisis: Composición por categorías en el tiempo

Gráfico generado: 

```python
frc = pd.crosstab(df['model_year'], df['cylinders'], normalize='index') * 100
plt.figure(figsize=(9, 5))
sns.lineplot(data=frc, markers=True, linewidth=2)
plt.title('Evolución de la Composición de Orígenes por Año del Modelo')
plt.xlabel('Año del Modelo (1970 - 1982)')
plt.ylabel('Proporción / Porcentaje (%)')
plt.legend(title='Origen', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
fr = pd.crosstab(df['model_year'], df['origin'], normalize='index') * 100

plt.figure(figsize=(9, 5))
sns.lineplot(data=fr, markers=True, linewidth=2)
plt.title('Evolución de la Composición de Orígenes por Año del Modelo')
plt.xlabel('Año del Modelo (1970 - 1982)')
plt.ylabel('Proporción / Porcentaje (%)')
plt.legend(title='Origen', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

¿Cómo ha evolucionado la participación de cada origen por año?
-> Estados Unidos dominó la producción desde 1970 y cayó significativamente en 1980, logrando recuperarse hasta 1982. En cambio, Japón y Europa han ido aumentando progresivamente su producción, sobre todo Japón en 1980. 

¿Disminuyó la presencia de vehículos con 8 cilindros con los años?

¿Qué relación encuentras entre este gráfico y la tendencia de mpg del Bloque I?

### BLOQUE 4

**¿Cómo se ve afectada la aceleración de los vehículos según la potencia del motor?**

Tipo de análisis: Relación entre dos variables cuantitativas 
Variables: horsepower (cuantitativa) y acceleration (cuantitativa)

Gráfico:

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='horsepower', y='acceleration', hue='origin', alpha=0.7, s=60)
sns.regplot(data=df, x='horsepower', y='acceleration', scatter=False, color='black', line_kws={'linestyle':'--', 'linewidth':2})
plt.title('Relación entre la Potencia y la Aceleración de los Vehículos')
plt.xlabel('Potencia (hp)')
plt.ylabel('Aceleración (0-60 mph en segundos)')
plt.legend(title='Origen')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

¿La relación entre la potencia y el tiempo de aceleración es positiva o negativa?

-> Es negativa. A mayor potencia , menor es el tiempo que tarda el vehículo en acelerar de 0 a 60 mph (es decir, el auto es más rápido).

¿La relación es lineal o muestra una curvatura?

-> Muestra una curvatura leve. En potencias bajas, un aumento pequeño de potencia reduce drásticamente el tiempo de aceleración. En cambio, en potencias muy altas, añadir más potencia ya no reduce el tiempo tan rápido.

¿Existen diferencias según el país de origen en esta relación?

-> Sí. Los autos estadounidenses ocupan la zona de alta potencia con tiempos de aceleración más bajos. Los autos japoneses y europeos se concentran en potencias bajas a medias, pero logran tiempos de aceleración competitivos.

**Interpretación técnica** El análisis de dispersión entre potencia y aceleración evidencia una relación inversamente proporcional. Es decir, a mayor potencia del motor, menor es el tiempo requerido. Se observa un rendimiento decreciente: incrementos de potencia en rangos bajos generan reducciones significativas en el tiempo de respuesta, mientras que en rangos superiores la curva tiende a aplanarse debido a la masa del vehículo y la pérdida de tracción.
