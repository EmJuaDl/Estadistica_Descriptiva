### Hoja de trabajo 2 Series simples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Carga de datos
ruta = 'Datos_tienda.csv'
datos = pd.read_csv(ruta)
```

#### Medidas de tendencia central

```python
promedio_ingreso= round(datos['Ingreso'].mean(),2)
```

```python
moda_ingreso = datos['Ingreso'].mode()[0]
```

```python
mediana_ingreso = round(datos['Ingreso'].median(),2)
print(f'-> El ingreso promedio de ventas es Q.{promedio_ingreso}.')
print(f'-> El ingreso que divide a las ventas en dos partes iguales es Q.{mediana_ingreso}.')
print(f'-> El ingreso más frecuente de ventas es Q.{moda_ingreso}.')
```

#### Medidas de dispersión

```python
ingresos= datos['Ingreso']
rango_ingresos = ingresos.max()-ingresos.min()
R_Ingresos = pd.cut(ingresos, rango_ingresos)
```

```python
desviacion_estandar = ingresos.std()
```

```python
varianza_ing = ingresos.var()
print(f'-> El rango de ingresos es Q.{rango_ingresos}.')
print(f'-> La desviación estándar de los ingresos es Q.{round(desviacion_estandar,2)}, una dispersión moderada.')
```

### Medidas de posición

```python
per75 = ingresos.quantile(0.75)
per20 = ingresos.quantile(0.20)
print(f'-> El ingreso que supera el 75% de la ventas es Q. {per75}.')
print(f'-> Los ingresos que delimitan el 20% inferior de las ventas es de entre Q. {ingresos.min()} y Q. {per20}.')

```
